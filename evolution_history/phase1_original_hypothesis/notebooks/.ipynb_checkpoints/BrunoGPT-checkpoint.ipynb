```python
import json
from pathlib import Path
from IPython.display import Markdown
```

```python
# External snapshot path
json_dir = Path(r"D:\gpt_log\GPT bugged conversation\Bruno\bugged_json")

# Load snapshots
snapshots = {}
for file in sorted(json_dir.glob("bugged_*.json")):
    with open(file, encoding="utf-8") as f:
        data = json.load(f)
        snapshots[data["snapshot"]] = data["lines"]
```

```python
display(Markdown("### 🔴 Snapshot Index\n" + "\n".join(
    f"- `{k}`: {len(v)} lines" for k, v in snapshots.items()
)))
```

```python
def view_snapshot(name, start=0, end=50):
    lines = snapshots.get(name, [])
    if not lines:
        print(f"❌ No snapshot named {name}")
        return
    print(f"📄 {name} — lines {start} to {end}\n")
    for i, line in enumerate(lines[start:end], start=start):
        print(f"{i:>5}: {line}")
```

```python
def search_snapshots(keyword):
    for name, lines in snapshots.items():
        for i, line in enumerate(lines):
            if keyword.lower() in line.lower():
                print(f"{name}:{i} → {line}")
```

```python
def reload_snapshots():
    snapshots.clear()
    for file in sorted(json_dir.glob("bugged_*.json")):
        with open(file, encoding="utf-8") as f:
            data = json.load(f)
            snapshots[data["snapshot"]] = data["lines"]
```

```python
import json
from datetime import datetime
from pathlib import Path

def update_context(snapshot_file="bugged_snapshots.json", output="bruno_gpt_context.json"):
    with open(snapshot_file, encoding="utf-8") as f:
        snapshots = json.load(f)

    context = {
        "project": "Bruno Entropy Collapse",
        "source_files": [
            "bugged_snapshots.json",
            "BrunoGPT.ipynb"
        ],
        "summary": {
            "scope": "Theoretical and numerical exploration of a thermodynamic collapse condition (Bruno Constant) across astrophysical systems.",
            "constant": "β_B = κ · T with κ ≈ 1366 K⁻¹",
            "collapse_criteria": "Entropy collapse when β_B ≥ 1 → V/A ≤ κ · T",
            "validation": [
                "SN1987A neutrino burst",
                "GW150914 pre-collapse phase",
                "White Dwarf and Neutron Star transitions",
                "Primordial black holes",
                "Consistency with Jacobson's thermodynamic gravity"
            ],
            "tools": [
                "Jupyter notebook for snapshot viewing and GPT integration",
                "JSON-structured GPT conversation logs",
                "Cross-reference to Jacobson, Verlinde, Bekenstein frameworks"
            ],
            "scripts": [
                "convert_bugged.py",
                "split_snapshot.py",
                "summary_snapshot.py",
                "Tools_Cleaner.ipynb"
            ],
            "open_goals": [
                "Formal derivation of Bruno threshold within Jacobson model",
                "Derive observational signatures (e.g., neutrino flux, GW entropy slope)",
                "Draft for preprint submission (e.g., arXiv)",
                "Cross-scale test of Bruno with PBH, CMB, GR simulations"
            ],
            "log_status": {
                "snapshot_count": len(snapshots),
                "total_lines": sum(len(e['lines']) for e in snapshots)
            },
            "last_updated": datetime.utcnow().isoformat() + "Z"
        }
    }

    with open(output, "w", encoding="utf-8") as f:
        json.dump(context, f, indent=2, ensure_ascii=False)

    print(f"✅ Context updated: {output} — {context['summary']['log_status']}")
```
